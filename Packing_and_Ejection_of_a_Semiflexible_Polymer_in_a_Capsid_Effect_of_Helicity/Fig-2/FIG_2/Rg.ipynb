```python
import numpy as np
import os
avgRg={}
z=0
k_phi={}
Rgerr={}
p=[0,10,30]
for i in range(3):              

                    k_phi[i]=[]
                    avgRg[i]=[]
                    Rgerr[i]=[]
                    # Read the file and store data in three lists
                    if not os.path.isfile(f'avgrg_kappa_theta{p[i]}_stderr.txt'):
                    
                        continue
                    z+=1 
                    
                    with open(f'avgrg_kappa_theta{p[i]}_stderr.txt', 'r') as file:
                        next (file)
                        for line in file:
                            line = line.strip()  # Remove leading/trailing whitespace
            
                            parts = line.split()
                           
                            if len(parts) == 3:
                            
                                 
                                k_phi[i].append(float(parts[0]))
                                avgRg[i].append(float(parts[1]))
                                Rgerr[i].append(float(parts[2]))                                                    
```

```python
import matplotlib.pyplot as plt
%config InlineBackend.figure_format = 'retina'
plt.rcParams.update({'font.size':24,
    "text.usetex": True,
    "font.family": "sans-serif",
    "font.sans-serif": "Helvetica",
})

plt.figure(figsize=(10,8))

plt.errorbar(
    k_phi[0], avgRg[0], Rgerr[0], 
    fmt="o-", color="red", ecolor="red",  
    capsize=4, capthick=1.5, elinewidth=1.5,  
    markersize=20, markerfacecolor="none", markeredgecolor="red", 
    linewidth=1.5, alpha=1,  
    rasterized=True,label=r'$\beta\kappa_\theta/\sigma=0$'
)
plt.errorbar(
    k_phi[1], avgRg[1], Rgerr[1],
    fmt="o-", color="blue", ecolor="blue",  
    capsize=4, capthick=1.5,   
    markersize=20, markerfacecolor="none", markeredgecolor="blue",  
     linewidth=1.5,alpha=1,  
    rasterized=True,label=r'$\beta\kappa_\theta/\sigma=10$'
)
plt.errorbar(
    k_phi[2], avgRg[2], Rgerr[2],
    fmt="o-", color="green", ecolor="green",  
    capsize=4, capthick=1.5,   
    markersize=20, markerfacecolor="none", markeredgecolor="green",  
     linewidth=1.5,alpha=1,  
     antialiased=False,   
    rasterized=True,label=r'$\beta\kappa_\theta/\sigma=30$'
)
plt.legend(fontsize=26)
plt.xlabel(r'$\beta\kappa_\phi $', fontsize=30, fontweight='bold',labelpad=15)
plt.ylabel(r'${\langle R_g \rangle}(\sigma)$', fontsize=30,fontweight='bold',labelpad=15)
plt.xticks([0, 5, 10, 15, 20, 25],fontsize=30)
plt.yticks(fontsize=30)

plt.tick_params( direction="in", length=10, width=1.5)

plt.tight_layout()

# Save the figure with high resolution
plt.savefig('avgRG_combine1.pdf',dpi=800)
plt.savefig('avgRG_combine1.png', dpi=600)
# Display the plot
plt.show()
```
